```python
from transformers import MarianMTModel, MarianTokenizer

# Sample input: a Hindi sentence
input_text = "मैं अपने ऑर्डर के बारे में पूछना चाहता हूँ"

# Choose the correct Helsinki model for the source language → English
model_name = "Helsinki-NLP/opus-mt-hi-en"  # 'hi' is Hindi

# Load tokenizer and model
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Tokenize and translate
tokens = tokenizer(input_text, return_tensors="pt", padding=True)
translated = model.generate(**tokens)
result = tokenizer.decode(translated[0], skip_special_tokens=True)

print("🔤 Translated Text:", result)
```

```python
from transformers import MarianMTModel, MarianTokenizer, pipeline

# STEP 1: Translate (Hindi to English)
def translate_to_english(input_text, source_lang="hi"):
    model_name = f"Helsinki-NLP/opus-mt-{source_lang}-en"
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model = MarianMTModel.from_pretrained(model_name)
    tokens = tokenizer(input_text, return_tensors="pt", padding=True)
    translated = model.generate(**tokens)
    result = tokenizer.decode(translated[0], skip_special_tokens=True)
    return result

# STEP 2: Classify the translated text
def classify_text(text):
    classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
    candidate_labels = ["spam", "question", "opinion", "complaint", "request"]
    result = classifier(text, candidate_labels)
    return result['labels'][0], dict(zip(result['labels'], result['scores']))

# STEP 3: Complete Pipeline
def process_input(input_text, source_lang="hi"):
    print("🌐 Input:", input_text)
    
    translated = translate_to_english(input_text, source_lang)
    print("🔤 Translated:", translated)
    
    top_label, all_scores = classify_text(translated)
    print("🏷️ Prediction:", top_label)
    print("📊 All Scores:", all_scores)
    
    return top_label

# 🧪 Example: Hindi input
process_input("मुझे अपने आदेश के बारे में पूछताछ करनी है", source_lang="hi")
```
